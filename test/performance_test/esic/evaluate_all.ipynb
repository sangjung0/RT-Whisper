```python
from pathlib import Path
```

```python
from sj_utils.file.json import load_json
```

```python
from util import evaluate
```

```python
SEED = 42
MAX_COUNT = 1
TEST_ALL = True
USE_TOKEN_SAVER_LOADER = True
USE_PROMPT = False
SOURCE = "/workspaces/dev/datasets/ESIC-v1.1/v1.1/dev"
STORAGE = "/workspaces/dev/storage/esic/"
HYPERPARAMETER = "/workspaces/dev/test/optimize/esic/hyperparameters/20250731/step1_16b-96k-3090"
OUTPUT_PATH = "/workspaces/dev/test/performance_test/esic/output/test"
ESIC_VAL = "/workspaces/dev/test/performance_test/esic/data/val.json"
```

```python
DESCRIPTION = """
20250731/step1_16b-96k-3090 테스트
RTX4070 테스트
새로운 노멀라이저 추가
세이브로더 사용
"""
```

```python
source = Path(SOURCE)
storage = Path(STORAGE)
hyperparameter_path = Path(HYPERPARAMETER)
output_path = Path(OUTPUT_PATH)
esic_val = Path(ESIC_VAL)

if not source.exists():
    raise FileNotFoundError(f"Source path does not exist: {source}")
if not hyperparameter_path.exists():
    raise FileNotFoundError(f"Hyperparameter path does not exist: {hyperparameter_path}")
if not esic_val.exists():
    raise FileNotFoundError(f"ESIC validation file does not exist: {esic_val}")
if output_path.exists():
    raise FileExistsError(f"Output path already exists: {output_path}")
```

```python
data_paths = [Path(p) for p in load_json(esic_val)[1]]
hyperparameters = list(hyperparameter_path.glob("*.yaml"))
```

```python
for idx, hyperparameter in enumerate(hyperparameters):
    print(idx, hyperparameter.name)
    output = output_path / f"{hyperparameter.stem}.json"

    evaluate(
        source,
        storage,
        output,
        DESCRIPTION,
        data_paths,
        models = ["rt_whisper"],
        test_all = TEST_ALL,
        max_count=MAX_COUNT,
        seed = SEED,
        use_save_loader=USE_TOKEN_SAVER_LOADER,
        use_prompt=USE_PROMPT,
        hyperparameter = hyperparameter
    )

print("done")
```
