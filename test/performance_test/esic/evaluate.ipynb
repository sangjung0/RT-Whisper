```python
from pathlib import Path
```

```python
from sj_ai_utils.datasets.esic_v1 import ESICv1Dataset
```

```python
from common_util import evaluate
```

```python
SEED = 42
SAMPLE_SIZE = -1
TEST_ALL = False
USE_TOKEN_SAVER_LOADER = True
USE_PROMPT = False
CHUNK_SIZE = 48000

ESIC_VAL = "/workspaces/dev/test/performance_test/esic/data/val.json"  # use in val

STORAGE = "/workspaces/dev/.storage/esic/"
HYPERPARAMETER = "/workspaces/dev/test/optimize/all/hyperparameters/20250905/3s/step1_3s-96k-cpm-hyper/005_373_0.yaml"
OUTPUT_PATH = "/workspaces/dev/test/performance_test/esic/output/20250905/005_373_0-each.json"
```

```python
DESCRIPTION = """
USE_TOKEN_SAVER_LOADER = True
USE_PROMPT = False
"""
```

```python
TEST_MODELS = ["rt_whisper"]
# TEST_MODELS = ["whisper", "rt_whisper", "whisper_streaming"]
```

```python
storage = Path(STORAGE)
hyperparameter_path = Path(HYPERPARAMETER)
output_path = Path(OUTPUT_PATH)
esic_val = Path(ESIC_VAL)

if not hyperparameter_path.exists():
    raise FileNotFoundError(f"Hyperparameter path does not exist: {hyperparameter_path}")
if not esic_val.exists():
    raise FileNotFoundError(f"ESIC validation file does not exist: {esic_val}")
if output_path.exists():
    raise FileExistsError(f"Output path already exists: {output_path}")
```

```python
dataset = ESICv1Dataset.load(esic_val).sample(SAMPLE_SIZE)
```

```python
evaluate(
    storage,
    output_path,
    DESCRIPTION,
    dataset,
    models = TEST_MODELS,
    test_all = TEST_ALL,
    seed = SEED,
    use_save_loader=USE_TOKEN_SAVER_LOADER,
    use_prompt=USE_PROMPT,
    hyperparameter = hyperparameter_path,
    chunk_size=CHUNK_SIZE
)
```
