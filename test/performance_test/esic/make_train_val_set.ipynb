```python
from pathlib import Path
from sklearn.model_selection import train_test_split
```

```python
from sj_ai_utils.datasets.esic_v1 import search_all_data
from sj_utils.file.json import JsonSaver
```

```python
DESCRIPTION = """
ESIC dev 기준으로 train, val set을 생성한것입니다.
"""
```

```python
SOURCE = "/workspaces/dev/datasets/ESIC-v1.1/v1.1/dev/"
TRAIN_OUTPUT = "/workspaces/dev/test/performance_test/esic/esic_train.json"
VAL_OUTPUT = "/workspaces/dev/test/performance_test/esic/esic_val.json"
RANDOM_STATE = 42
```

```python
src = Path(SOURCE)
src.exists()
train_output = Path(TRAIN_OUTPUT)
val_output = Path(VAL_OUTPUT)

if train_output.exists():
    print(f"Train output file {train_output} already exists.")
if val_output.exists():
    print(f"Validation output file {val_output} already exists.")
```

```python
data_paths = search_all_data(src)
len(data_paths)
```

```python
train_paths, val_paths = train_test_split(data_paths, test_size=0.3, random_state=RANDOM_STATE)
print(len(train_paths), len(val_paths))
```

```python
train_paths = [str(p) for p in train_paths]
val_paths = [str(p) for p in val_paths]
```

```python
json_saver = JsonSaver(DESCRIPTION)
```

```python
json_saver.save(train_paths, train_output)
json_saver.save(val_paths, val_output)
```
