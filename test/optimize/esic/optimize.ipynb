```python
from pathlib import Path
```

```python
from sj_ai_utils.datasets.esic_v1 import ESICv1Dataset
```

```python
from rt_whisper_optimizer import Optimizer
```

```python
STORAGE = "/workspaces/dev/.storage/esic/"
STUDY = "/workspaces/dev/test/optimize/esic/study/"
ESIC_TRAIN = "/workspaces/dev/test/performance_test/esic/data/train.json"
CACHE = "/workspaces/dev/test/optimize/esic/.cache/20250826_cache.pkl"
```

```python
OUTPUT = [
    "/workspaces/dev/test/optimize/esic/hyperparameters/20250826/step1_16b-96k",
    "/workspaces/dev/test/optimize/esic/hyperparameters/20250826/step1_16b-96k-16cs",
    "/workspaces/dev/test/optimize/esic/hyperparameters/20250826/step1_16b-112k",
    "/workspaces/dev/test/optimize/esic/hyperparameters/20250826/step1_16b-112k-16cs",
]
STUDY_INSTRUCTION = [
    "/workspaces/dev/test/optimize/esic/study_param/20250826/step1_16b-96k.yaml",
    "/workspaces/dev/test/optimize/esic/study_param/20250826/step1_16b-96k-16cs.yaml",
    "/workspaces/dev/test/optimize/esic/study_param/20250826/step1_16b-112k.yaml",
    "/workspaces/dev/test/optimize/esic/study_param/20250826/step1_16b-112k-16cs.yaml",
]
```

```python
storage = Path(STORAGE)
study = Path(STUDY)
train = Path(ESIC_TRAIN)
cache = Path(CACHE)
if not train.exists():
    raise FileNotFoundError(f"Train file not found: {train}")
```

```python
datasets = ESICv1Dataset.load(train)
```

```python
for output_path, study_instruction in zip(OUTPUT, STUDY_INSTRUCTION):
    output = Path(output_path)
    study_instruction = Path(study_instruction)
    esic_optimizer = Optimizer(datasets, study, output, param_cache_path=cache, cache_storage=storage)
    esic_optimizer.optimize(study_instruction)
```
