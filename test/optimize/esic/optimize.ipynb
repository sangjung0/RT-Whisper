```python
from pathlib import Path
```

```python
from sj_ai_utils.datasets.esic_v1 import ESICv1Dataset
```

```python
from rt_whisper_optimizer import Optimizer
```

```python
ESIC_TRAIN = "/workspaces/dev/test/performance_test/esic/data/train.json"

STORAGE = "/workspaces/dev/.storage/esic/3s"
STUDY = "/workspaces/dev/test/optimize/esic/study/"
```

```python
OUTPUT = [
    "/workspaces/dev/test/optimize/esic/hyperparameters/20250917/3s/step2_3s-96k-cpm-2",
]
STUDY_INSTRUCTION = [
    "/workspaces/dev/test/optimize/esic/study_param/20250917/3s/step2_3s-96k-cpm.yaml",
]
```

```python
storage = Path(STORAGE)
study = Path(STUDY)
train = Path(ESIC_TRAIN)
if not train.exists():
    raise FileNotFoundError(f"Train file not found: {train}")
```

```python
datasets = ESICv1Dataset.load(train)[:10]
len(datasets)
```

```python
for output_path, study_instruction in zip(OUTPUT, STUDY_INSTRUCTION):
    output = Path(output_path)
    study_instruction = Path(study_instruction)
    optimizer = Optimizer(study, output, study_instruction, cache_storage=storage)
    optimizer.optimize(datasets, log_step = 100)
```
