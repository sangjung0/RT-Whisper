```python
import librosa
from faster_whisper import WhisperModel, BatchedInferencePipeline
```

```python
SAMPLE_RATE= 16000
MODEL_SIZE = "large-v3"
```

```python
AUDIO_FILE_1 = "/workspaces/dev/.data/news_with_english.mp3"
AUDIO_FILE_2 = "/workspaces/dev/.data/news_with_russian.mp3"
```

```python
audio_1, sr_1 = librosa.load(AUDIO_FILE_1, sr=SAMPLE_RATE)
audio_2, sr_2 = librosa.load(AUDIO_FILE_2, sr=SAMPLE_RATE)
```

```python
model = WhisperModel(MODEL_SIZE, device="cuda", compute_type="int8")
```

```python
segments, info = model.transcribe(
    [audio_1, audio_2],
    beam_size=2,
    word_timestamps=True,
    vad_filter=True,
)

for segment in segments:
    # print("[%.2fs -> %.2fs] %s" % (segment.start, segment.end, segment.text))
    print(segment)
```
